In order to assess the performance of the model we will use the top 1, top 3, and top 5 accuracy, as well as the macro f1 score. We use the macro f1 score since the dataset is heavily imbalanced, and we do not want the f1 score to be dominated be classes with a higher number of samples.

We will assess both models, one of which was trained with a random split, and the other with a stratified split, over ten epochs.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from src.clip_benchmark import benchmark_clip

checkpoints = sorted(Path("model/mushroomCLIP_randomSplit").glob("checkpoint-*"))

rows = []
per_class_rows = []

for epoch, ckpt in enumerate(checkpoints, start=1):
    metrics = benchmark_clip(
        adapter_path=str(ckpt),
        topk_values=(1, 3, 5),
        include_macro_f1=True,
        include_per_class_f1=True,
        show_progress=False,
    )

    rows.append({
        "epoch": epoch,
        "top1 acc.": metrics["top1_accuracy"],
        "top3 acc.": metrics["top3_accuracy"],
        "top5 acc.": metrics["top5_accuracy"],
        "macro_f1": metrics["macro_f1"],
    })

    for cls, vals in metrics["per_class"].items():
        per_class_rows.append({
            "epoch": epoch,
            "class": cls,
            "f1": vals["f1"] * 100.0,
        })

df_random = pd.DataFrame(rows).set_index("epoch")
pc_random = pd.DataFrame(per_class_rows)
```

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from src.clip_benchmark import benchmark_clip

checkpoints = sorted(Path("model/mushroomCLIP_stratifiedSplit").glob("checkpoint-*"))

rows = []
per_class_rows = []

for epoch, ckpt in enumerate(checkpoints, start=1):
    metrics = benchmark_clip(
        adapter_path=str(ckpt),
        topk_values=(1, 3, 5),
        include_macro_f1=True,
        include_per_class_f1=True,
        show_progress=False,
    )

    rows.append({
        "epoch": epoch,
        "top1 acc.": metrics["top1_accuracy"],
        "top3 acc.": metrics["top3_accuracy"],
        "top5 acc.": metrics["top5_accuracy"],
        "macro_f1": metrics["macro_f1"],
    })

    for cls, vals in metrics["per_class"].items():
        per_class_rows.append({
            "epoch": epoch,
            "class": cls,
            "f1": vals["f1"] * 100.0,
        })

df_stratified = pd.DataFrame(rows).set_index("epoch")
pc_stratified = pd.DataFrame(per_class_rows)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

fig, axes = plt.subplots(1, 2, figsize=(16, 5), sharey=True)

for ax, df_local, title in [
    (axes[0], df_random, "Random Split"),
    (axes[1], df_stratified, "Stratified Split"),
]:
    best_epoch = df_local["macro_f1"].idxmax()
    best_value = df_local.loc[best_epoch, "macro_f1"]

    df_local[["top1 acc.", "top3 acc.", "top5 acc.", "macro_f1"]].plot(marker="o", ax=ax)

    ax.axvline(best_epoch, color="red", linestyle="--", alpha=0.7, label=f"best epoch = {best_epoch}")
    ax.scatter([best_epoch], [best_value], color="red", zorder=5)
    ax.annotate(
        f"best macro_f1: {best_value:.2f}%",
        xy=(best_epoch, best_value),
        xytext=(best_epoch + 0.2, best_value + 0.5),
        arrowprops=dict(arrowstyle="->", color="red"),
        color="red",
        fontsize=9,
    )
    ax.set_title(f"{title}: Validation Metrics by Epoch")
    ax.set_xlabel("Epoch")
    ax.set_ylabel("%")
    ax.grid(True, alpha=0.3)
    ax.legend()

plt.tight_layout()
plt.show()
```
